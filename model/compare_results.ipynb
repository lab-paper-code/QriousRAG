```python
import pandas as pd
import os

res_df = pd.DataFrame(columns=['model','timestamp','rougeLsum','length','str_em','ovscore'])

for file in os.listdir('./results'):
    if file.endswith('.csv'):
        mean = pd.read_csv(f'./results/{file}')
        mean=mean[mean['length']<1000][:97].mean().to_list()
        file_name  = file.strip('.csv').split('_')
        model_name = file_name[0] if len(file_name) >=1 else 'unk'
        timestamp = pd.to_datetime(file_name[2] if len(file_name) >= 3 else None, format="%Y%m%d-%H%M%S")
        res_df.loc[len(res_df)] = [file_name, timestamp] + mean

res_df.sort_values('timestamp', ascending=False)
```
