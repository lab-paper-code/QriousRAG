```python
import re
from bs4 import BeautifulSoup
#parse tables to text
def get_table(table):
    table_text = []
    
    # tbody가 없을 경우 table 내에서 바로 tr 태그를 찾음
    tbody = table.find('tbody')
    if tbody:
        rows = tbody.findChildren("tr", recursive=False)
    else:
        rows = table.find_all("tr")  # tbody가 없을 경우 직접 tr 태그를 찾음
    
    for i, tr in enumerate(rows):
        tr_text = tr.get_text()
        tr_text = re.sub(r'\n+',';',tr_text).strip(';')
        if not tr_text:
            continue
        if table_text == []:
            tr_text = '\n#### Table: ' + tr_text
        table_text.append(tr_text)

    return '\n'.join(table_text)

# parse pars to text
def get_p(par):
    p_text = par.get_text()
    p_text = p_text.replace('\n','')
    return p_text

def get_h(heading):
    h = heading.find(['h1', 'h2', 'h3','h4', 'h5'])
    try:
        heading_type = int(re.search(r'<h(\d)', str(h)).group(1))
    except:
        print(heading)
        raise
    h_text ='\n' + ' '.join(['#'*heading_type,h.get_text()])
    return h_text

#pars unordered lists to text
def get_ul(ul):
    list_text = []
    for li in ul.find_all('li'):
        list_text.append('* ' + li.get_text())
    return '\n'.join(list_text)

# pars ordered list to text
def get_ol(ol):
    list_text = []
    for i, li in enumerate(ol.find_all('li')):
        if li.get_text():
            list_text.append(' '.join([str(i+1),li.get_text().replace('\n','')]))
    return '\n'.join(list_text)

# parse the whole document
def parse_document(doc):
    soup = BeautifulSoup(doc, 'html.parser')  # BeautifulSoup 객체로 변환
    content = soup.find_all(['div', 'p', 'table', 'ul', 'ol'])  # 원하는 태그들 추출
    document  = []
    for cont in content:
        # stop condition
        if cont.name == 'div' and cont.find(['h1', 'h2', 'h3','h4', 'h5'], id=['See_also', 'References']):
            break
        
        # get heading
        if cont.name == 'div' and cont.has_attr('class') and  'mw-heading' in cont['class']:
            document.append(get_h(cont))
        # get paragraph
        elif cont.name == 'p':
            par = get_p(cont)
            if par:
                document.append(par)
        # get unordered list
        elif cont.name == 'ul':
            document.append(get_ul(cont))
        # get ordered list
        elif cont.name == 'ol':
            document.append(get_ol(cont))
        # get table
        elif cont.name == 'table':
            if cont.has_attr('class') and 'metadata' in cont['class']: 
                continue
            document.append(get_table(cont))
        # explore div
        elif cont.name == 'div':
            document.append(parse_document(cont))
    return ' '.join(document)  # 텍스트 합치기
```

```python
# create evidence files
files = {}
for s in sample:
    id = s['question_text']
    text = s['document_text'].split()  # 문서를 단어 단위로 나누기
    chunks = [(chunk['start_token'], chunk['end_token']) for chunk in s['long_answer_candidates'] if chunk['top_level'] == True]
    
    for start, end in chunks:
        print(f"Processing chunk: start={start}, end={end}")
        # 수정된 슬라이싱 구문
        parsed_doc = parse_document(' '.join(text[start:end]))
        if id not in files:
            files[id]=list()
        files[id].append(parsed_doc)  # 올바른 슬라이싱 사용

print(f"Total evidence files created: {len(files)}")
    
    
```

```python
nq_data=list(files.items())
nq_data[0]
```

```python
# npy 파일로 저장
save_list_as_npy(nq_data, '/raid/deallab/SF_RAG_Data/N_Q/nq_data.npy')
```
